```python
import numpy 
import scipy.io
import math
import geneNewData
```

```python
### TEST FUNCTION: test_question1
myID = '2599'
geneNewData.geneData(myID)

Numpyfile0 = scipy.io.loadmat('digit0_stu_train' + myID + '.mat')
Numpyfile1 = scipy.io.loadmat('digit1_stu_train' + myID + '.mat')
Numpyfile2 = scipy.io.loadmat('digit0_testset' + '.mat')
Numpyfile3 = scipy.io.loadmat('digit1_testset' + '.mat')

train0 = Numpyfile0.get('target_img')   
train1 = Numpyfile1.get('target_img')   
test0  = Numpyfile2.get('target_img')   
test1  = Numpyfile3.get('target_img')   

def extract_features(images):
    if images.shape[0] == 28:
        N = images.shape[2]
        flat = images.reshape(784, N).T.astype(float)
    else:
        N = images.shape[0]
        flat = images.reshape(N, 784).astype(float)

    feat1 = flat.mean(axis=1)
    feat2 = flat.std(axis=1)

    return numpy.column_stack([feat1, feat2])    

train0_feat = extract_features(train0)
train1_feat = extract_features(train1)
test0_feat  = extract_features(test0)
test1_feat  = extract_features(test1)

# Digit 0
mean_f1_d0 = train0_feat[:, 0].mean()   
var_f1_d0  = train0_feat[:, 0].var()    
mean_f2_d0 = train0_feat[:, 1].mean()   
var_f2_d0  = train0_feat[:, 1].var()    

# Digit 1
mean_f1_d1 = train1_feat[:, 0].mean()   
var_f1_d1  = train1_feat[:, 0].var()    
mean_f2_d1 = train1_feat[:, 1].mean()   
var_f2_d1  = train1_feat[:, 1].var()    

def log_gaussian(x, mean, var):
    return -0.5 * math.log(2 * math.pi * var) - ((x - mean) ** 2) / (2 * var)

def naive_bayes_predict(features):
    predictions = []
    for i in range(len(features)):
        f1 = features[i, 0]
        f2 = features[i, 1]

        log_p0 = log_gaussian(f1, mean_f1_d0, var_f1_d0) + \
                    log_gaussian(f2, mean_f2_d0, var_f2_d0)

        log_p1 = log_gaussian(f1, mean_f1_d1, var_f1_d1) + \
                    log_gaussian(f2, mean_f2_d1, var_f2_d1)

        predictions.append(0 if log_p0 >= log_p1 else 1)

    return numpy.array(predictions)

pred_test0 = naive_bayes_predict(test0_feat)   
pred_test1 = naive_bayes_predict(test1_feat)   

accuracy_digit0 = numpy.mean(pred_test0 == 0)
accuracy_digit1 = numpy.mean(pred_test1 == 1)

result = [
        myID,
        mean_f1_d0, var_f1_d0,
        mean_f2_d0, var_f2_d0,
        mean_f1_d1, var_f1_d1,
        mean_f2_d1, var_f2_d1,
        accuracy_digit0,
        accuracy_digit1
    ]

result = [myID] + [float(x) for x in result[1:]]

print(result)
```
